### 5. Выявите основных работодателей, предлагаемые зарплаты и условия труда для аналитиков.

```python
# 1. Агрегация зарплат и количества вакансий
summary = (
    df.groupby('employer')
      .agg(
          vacancies_count=('id', 'count'),
          salary_from_mean=('salary_from', 'mean'),
          salary_to_mean=('salary_to', 'mean')
      )
      .round(2)
)

# 2. Самые частые значения (mode)
schedule_mode = df.groupby('employer')['schedule'].agg(pd.Series.mode).groupby('employer').first()
employment_mode = df.groupby('employer')['employment'].agg(pd.Series.mode).groupby('employer').first()
experience_mode = df.groupby('employer')['experience'].agg(pd.Series.mode).groupby('employer').first()

# 3. Объединение в одну таблицу
summary['schedule'] = schedule_mode
summary['employment'] = employment_mode
summary['experience'] = experience_mode

# 4. Сортировка и топ-10
summary = summary.sort_values('vacancies_count', ascending=False).head(10)

summary
```

Основными работодателями по количеству вакансий являются СБЕР, WILDBERRIES, Ozon, Банк ВТБ и T1. Лидером является СБЕР — он разместил 243 вакансии, что значительно больше, чем у остальных компаний.
По зарплатам данные представлены не для всех работодателей: у части компаний значения salary_from_mean и salary_to_mean отсутствуют. Среди компаний с указанной зарплатой наиболее высокие средние значения наблюдаются у WILDBERRIES: нижний порог — 180 000 руб., верхний — 220 000 руб.
По условиям труда чаще всего встречаются полная занятость и график полный день. Исключением среди топ-компаний является МАГНИТ, где чаще встречается удалённая работа.
По опыту большинство вакансий рассчитано на специалистов уровня Junior+ / Middle, чаще всего с опытом от 1 до 3 лет или от 3 до 6 лет.

### 6. Определите наиболее востребованные навыки (как жёсткие, так и мягкие) для различных грейдов и позиций.

```python
key_skills = (
    df.melt(
        id_vars='experience',
        value_vars=['key_skills_1','key_skills_2','key_skills_3','key_skills_4'],
        value_name='skill'
    )
    .dropna()
)

key_skills_top = (
    key_skills
    .groupby(['experience', 'skill'])
    .size()
    .reset_index(name='count')
    .sort_values(['experience', 'count'], ascending=[True, False])
    .groupby('experience')
    .head(5)
)

key_skills_top = key_skills_top.reset_index(drop=True)
key_skills_top
```

```python
soft_skills = (
    df.melt(
        id_vars='experience',
        value_vars=['soft_skills_1','soft_skills_2','soft_skills_3','soft_skills_4'],
        value_name='skill'
    )
    .dropna()
)

soft_skills_top = (
    soft_skills
    .groupby(['experience', 'skill'])
    .size()
    .reset_index(name='count')
    .sort_values(['experience', 'count'], ascending=[True, False])
    .groupby('experience')
    .head(5)
)

soft_skills_top = soft_skills_top.reset_index(drop=True)
soft_skills_top
```

Наиболее востребованными жёсткими навыками для аналитиков являются SQL, Python и анализ данных. Эти навыки встречаются на всех уровнях: от Junior до Senior, что говорит об их базовой важности для профессии аналитика.
Для уровня Junior чаще всего требуются SQL, Python и анализ данных. Для Junior+ и Middle набор навыков похожий, но количество вакансий значительно выше: особенно часто встречаются SQL, Python, анализ данных и документация. Для Senior также лидируют SQL и Python, но выборка по этому грейду небольшая, поэтому выводы по нему нужно делать осторожно.
Среди мягких навыков чаще всего встречаются документация, коммуникация и аналитическое мышление. Для более опытных специалистов также встречаются навыки презентации результатов и проактивность.
Важно отметить, что во многих вакансиях навыки не указаны, поэтому значение «Не указан» занимает первое место почти во всех грейдах. Поэтому при интерпретации лучше смотреть на реальные навыки после исключения этого значения.
